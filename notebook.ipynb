# Resume Analyzer

Compares a resume (`.pdf` or `.docx`) against a job posting URL using Claude.
Outputs match scores, skill gaps with reasoning, resume edits, and a prioritized action list.

**Setup:**
1. Copy `.env.example` → `.env` and set `ANTHROPIC_API_KEY`
2. Install dependencies: `pip install -r requirements.txt`
3. Install Playwright browser *(optional — needed for JS-heavy job boards)*: `playwright install chromium`
4. Run the cell below

**Scraping notes:**
- Most public job boards (Indeed, Greenhouse, Lever, iCIMS) work via direct HTTP fetch.
- JS-rendered boards (React/Vue career pages) automatically fall back to Playwright if installed.
- Workday is not supported — paste the description into the *Job text* field instead.
- LinkedIn requires login — paste the description instead.
